# Lista nativa

`````python
notas = [55, 68, 72, 49, 80]
`````

Aquí hay una lista nativa de Python.

---
**¿Qué ocurre si quiero sumar 5 puntos a todas las notas?**

---
Con una lista, normalmente se puede hacer:

`````python
notas_nuevas = []

for nota in notas:
  notas_nuevas.append(nota + 5)

print(notas_nuevas)
`````

**Concepto**

```text
Las listas de Python son estructuras generales.
No están diseñadas específicamente para realizar operaciones matemáticas masivas sobre datos numéricos.
```

Esta línea de código
```python
notas + 5
```

va a generar el error:

```
TypeError: can only concatenate list (not "int") to list
```

# Pasar de lista a NumPy

`````python
notas = np.array([55, 68, 72, 49, 80])
`````

`````python
notas_nuevas = notas + 5

print(notas_nuevas)
`````

Lo anterior es lo que se conoce como `Vectorización`, ya no es necesario hacer:

```python
for nota in notas:
  ...
```

**Importante**

**Vectorizar** no significa que internamente no exista ningún ciclo.

Significa que el ciclo de procesamiento está implementado en código optimizado de NumPy, generalmente en C, en lugar de ejecutarse elemento por elemento mediante el intérprete de Python.

## `ndarray`

Un ndarray está pensado para trabajar con datos homogéneos:

```python
datos = np.array([10, 20, 30, 40])
```

Conceptualmente:
```text
┌────┬────┬────┬────┐
│ 10 │ 20 │ 30 │ 40 │
└────┴────┴────┴────┘
      memoria contigua
```

Todos los elementos tienen un tipo común:

```python
datos.dtype
```

retornará `int64`

`````python
datos = np.array([10, 20, 30, 40])
datos.dtype
`````

# El concepto crítico: `.loc` vs `.iloc`


`````python
data = pd.DataFrame({
    "nombre": ["Anacleto", "Mefistofeles", "Pericles"],
    "edad": [20, 25, 30],
    "nota": [6.1, 5.5, 6.8]
})
data
`````

## `.iloc`: posición

`````python
data.iloc[0]
`````

`iloc` significa, conceptualmente:
```text
integer location
```
Trabaja con posiciones enteras.

`````python
data.iloc[0, 1]
`````

## `.loc`: etiqueta

`````python
data.loc[0]
`````

En cambio:
```python
data.loc[0]
```
significa:
```
Dame la fila cuya etiqueta de índice es 0.
```

Al principio parecen exactamente iguales:

```python
data.iloc[0]
data.loc[0]
```

Sin embargo, **SON DISTINTOS**

`````python
data.index = ["A", "B", "C"]
`````

`````python
data
`````

`````python
data.loc["A"]
`````

`````python
data.iloc[0]
`````

Sin embargo

```python
data.loc[0]
```

Marcará error:
```text
KeyError: 0
``

`````python
data = pd.DataFrame(
    {
        "nombre": ["Ana", "Juan", "Pedro"],
        "nota": [6.1, 5.5, 6.8]
    },
    index=[101, 205, 310]
)
data
`````

`````python
data.loc[205]
`````

`````python
data.iloc[1]
`````

```python
data.iloc[205]
```

marcará error porque intentará buscar el elemento ubicado en la posición 205

y

```python
data.loc[1]
```

también marcará error porque busca una etiqueta 1 que tampoco existe.
